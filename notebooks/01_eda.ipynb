```python
import pandas as pd
import numpy as np
df = pd.read_csv("C:/Users/HP/Desktop/credit_card_fraud_detection/data/creditcard.csv")
print(df.shape)
df.head()
missing_values = df.isnull().sum()
print("------Missing Values Per column----");
print(missing_values)
print("\n--Total missing values in the entire dataset--")
print(missing_values.sum())
duplicates = df.duplicated().sum()
print(f"Total no of duplicate rows:{duplicates}")
df.drop_duplicates(inplace =True)
print(df.shape)
class_count =  df['Class'].value_counts()
legit_count = class_count[0]
fraud_count = class_count[1]
fraud_percentage = (fraud_count/len(df)) * 100
print("--- Class Distribution ---")
print(f"Legitimate (Class 0): {legit_count:,}")
print(f"Fraudulent (Class 1): {fraud_count:,}")
print(f"Fraud Percentage: {fraud_percentage:.3f}%")
df.head
print("-------Transaction Amount Summry Statistics-----")
amount_summary = df.groupby("Class")["Amount"].describe()
print(amount_summary)
corr = df.corr()["Class"].sort_values()
print("---Top 5 Most Negatively Correlated Features with Fraud---")
print(corr.head(5))
print("\n---Top 5 Most Positively Correlated Features with Fraud---")
print(corr.tail(6))
```
